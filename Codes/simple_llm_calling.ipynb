```python
import os
from dotenv import load_dotenv
from langchain_groq import ChatGroq
load_dotenv()

llm = ChatGroq(
    model="llama-3.1-8b-instant",
    temperature=0.7
)
```

```python
with open(r"C:\Users\Pratik\OneDrive\Documents\Rag Projects\Healthcare\DATA\blood_work.txt", "r") as f:
    blood_report = f.read()

print(blood_report[:200])
```

```python
prompt = f"""
You are a helpful assistant for doctors. 
From the blood report below,extract all the test values and classify each one as HIGH. LOW, or NORMAL
based on the references ranges provided in the report.and

Format your response as:
- Test Name: value| Status: High/Low/Normal | Reference: range

Here is the blood report:{blood_report}
"""

extracted_response= llm.invoke(prompt)
extracted_values = extracted_response.text
print("=== STAGE 1: EXTRACTED VALUES === \n ")
print(extracted_values)
```
